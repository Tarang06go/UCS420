## 2. Steps Data and Aggregations

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])
```

```python
total = np.sum(steps)
print('Total:', total)

mean = np.mean(steps)
print('Mean:', mean)
```

```python
maxi = np.max(steps)
mini = np.min(steps)

print(maxi)
print(mini)
```

```python
print(np.sum(steps, axis=0))
print(np.sum(steps, axis=1))
```

## 3. NumPy Slicing, Views, Copies, and Array Properties

```python
original = np.array([1, 2, 3, 4, 5, 6])
print('Original:', original)

subset = original[1:5]
print('Subset:', subset)

subset[0] = 999

print(original)
print(subset)

copied = original[1:5].copy()
copied[0] = 500

print('Original:', original)
print('Copied:', copied)

matrix = np.arange(1, 13).reshape(3, 4)
print(matrix)

first_row = matrix[0, :]
last_row = matrix[-1, :]
second_column = matrix[:, 1]
selected = matrix[1:3, 2:4]

print(first_row)
print(last_row)
print(second_column)
print(selected)

flat = matrix.flatten()
rav = matrix.ravel()

print(flat)
print(rav)

rav[0] = 1000
print(rav)
print(matrix)

flat[1] = 2000
print(flat)
print(matrix)

print(matrix.shape)
print(matrix.ndim)
print(matrix.size)
print(matrix.dtype)
```

## 4. Matrix Operations and Linear Regression

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print('Shape of X:', X.shape)
print('Dimensions of X:', X.ndim)

XT = X.T
print('\nX.T:')
print(XT)

XTX = X.T @ X
print('\nX.T @ X:')
print(XTX)

inverse = np.linalg.inv(XTX)
print('\nInverse of X.T @ X:')
print(inverse)

beta = np.linalg.inv(X.T @ X) @ X.T @ y
print('\nCoefficients:')
print(beta)

print('\nInterpretation:')
print('Sleep coefficient:', beta[0])
print('Activity coefficient:', beta[1])
print('Stress coefficient:', beta[2])

new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta

print('\nPredicted assistance score:', predicted_score)
```
