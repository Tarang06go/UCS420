```python
# A6.2
def isprime(n):
    s = 0

    for i in range(2,n + 1):
        prime = True
    for j in range(2, i):
      if i % j == 0:
          prime = False
          break
      if prime:
        s += i
    return s

n = int(input("Enter n: "))
print(isprime(n))
```
