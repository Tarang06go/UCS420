```python
import io
import PIL
from PIL import Image

print("Pillow version:", PIL.__version__)

# Test Pillow's JPEG encoder independently of LandingAI
buffer = io.BytesIO()
test_image = Image.new("RGB", (10, 10), color="red")
test_image.save(buffer, format="JPEG")

print("JPEG encoding successful:", len(buffer.getvalue()), "bytes")
```

```python

from google.colab import userdata
from landingai.predict import Predictor

API_KEY = userdata.get("LANDINGAI_API_KEY")

ENDPOINT_ID = "c5108097-6a1f-4347-8f1b-210917136468"

predictor = Predictor(
    endpoint_id=ENDPOINT_ID,
    api_key=API_KEY
)

print("LandingAI predictor initialized.")
```

```python

from google.colab import files
from PIL import Image
import io
import numpy as np

uploaded = files.upload()
filename = next(iter(uploaded))

with Image.open(io.BytesIO(uploaded[filename])) as im:
    image = im.convert("RGB").copy()

image_array = np.array(image)

print("Image size:", image.size)
print("Image array shape:", image_array.shape)
```

```python
predictions = predictor.predict(image_array)

print("Prediction count:", len(predictions))

for i, prediction in enumerate(predictions, start=1):
    print(f"\nPrediction {i}:")
    print(prediction)
```

```python

from gtts import gTTS
from IPython.display import Audio, display

if len(predictions) > 0:
    predicted_label = predictions[0].label_name
    if predicted_label == "Face":
        result = "Face detected."
    else:
        result = f"Detected {predicted_label}."
else:
    result = "No face detected."

print("Spoken output:", result)
tts = gTTS(text=result, lang="en")
tts.save("output.mp3")
display(Audio("output.mp3", autoplay=True))
```
